```python
import random
```

```python
r = random.randint(0, 2)
if r == 0:
    print("goo")
elif r == 1:
    print("choki")
elif r == 2:
    print("per")
```

```python
match r:
    case 0:
        print("goo")
    case 1:
        print("chiki")
    case 2:
        print("per")
```
