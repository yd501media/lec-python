# **Errors and Exceptions**
Pythonでは、プログラムの実行中にエラーや例外が発生することがあります。  
これらを適切に処理することで、プログラムが中断するのを防ぎ、ユーザーにとって使いやすいアプリケーションを作ることができます。

# エラーと例外
Pythonでは、プログラムの実行中に問題が発生すると「エラー」または「例外」として報告されます。
  
- **エラー**: プログラムに文法的な間違いがあり、コードの実行が始まる前に検出される問題。例: 構文エラー。
- **例外**: プログラムの実行中に発生する問題。例: データ型の不一致やゼロ除算など。
  
Pythonでは、例外が発生するとプログラムが中断されますが、適切な処理を行うことで例外を制御し、プログラムの継続が可能です。

### 構文エラーの例
構文エラーは、プログラムの実行が開始される前に検出されるエラーです。

```python
# 構文エラー: 文法が間違っている場合
print("Hello   # 終了のクォートが不足している
```

### 実行時例外の例
実行時例外は、コードの実行中に発生するエラーで、未処理の場合プログラムが中断されます。

```python
# 実行時例外の例
x = 10 / 0  # ゼロで除算しようとする
```

# 例外処理
例外処理とは、プログラムの実行中に発生するエラー（例外）を適切に制御し、プログラムの中断を防ぐ仕組みです。  
Pythonでは、`try-except` 構文を使用して例外をキャッチし、特定のエラーに応じた処理を実行できます。

### 基本的な例外処理
`try` ブロック内で例外が発生した場合、`except` ブロックで適切に処理することで、プログラムの中断を防ぎます。

```python
# try-exceptを使用した例外処理
try:
    x = 10 / 0  # ここで例外が発生
except ZeroDivisionError:
    print("You cannot divide by zero!")
```

### 複数の例外を処理
複数の例外を個別に処理することで、異なるエラーに対して適切な対処が可能です。

```python
# 複数の例外に対応する処理
try:
    x = int("abc")  # ValueError が発生
except ZeroDivisionError:
    print("Cannot divide by zero!")
except ValueError:
    print("Invalid input: cannot convert to integer.")
```

### 全ての例外を処理
`except Exception` を使用することで、全ての例外をキャッチ可能です。  
以下の例では、発生した例外の詳細情報を変数 `e` に格納し、表示します。

```python
try:
    # 任意の例外が発生する可能性のある処理
    x = int("abc")
except Exception as e:
    # 全ての例外をキャッチ
    print(f"An error occurred: {e}")
```

### `else` を使用した例外処理
`else` ブロックは例外が発生しなかった場合にのみ実行されます。

```python
try:
    x = int("123")  # 正常に実行
except ValueError:
    print("Invalid input!")
else:
    print(f"Conversion succeeded: {x}")
```

### `finally` を使用した例外処理
`finally` ブロックは、例外の有無にかかわらず必ず実行されるコードを記述するために使用されます。

```python
try:
    x = int("abc")
except ValueError:
    print("Invalid input!")
finally:
    print("Execution completed.")
```

# カスタム例外
カスタム例外とは、Pythonの組み込み例外（例: `ValueError`, `TypeError`）では対応しきれない特定のエラー状況を扱うために、自分で定義する例外クラスです。  
独自の例外を定義することで、エラーの意味を明確にし、エラーハンドリングをより分かりやすくすることができます。  
Pythonでは、カスタム例外を作成する際に、組み込みクラス `Exception` を継承して新しい例外クラスを作成します。

### カスタム例外の基本例

```python
# カスタム例外クラスの定義
class CustomError(Exception):
    """カスタム例外の基本クラス"""
    pass

# カスタム例外の使用
try:
    raise CustomError("This is a custom error!")
except CustomError as e:
    print(f"Caught custom exception: {e}")
```

# 例外の種類
Pythonでは、さまざまな種類の組み込み例外が用意されています。  
それぞれ特定のエラー状況を表現し、適切なエラーハンドリングを行うために使用されます。
  
以下に、代表的な例外とその発生条件を示します:  

- `ValueError`: 無効な値が渡された場合。
- `TypeError`: 不適切なデータ型が使用された場合。
- `ZeroDivisionError`: 0で割ろうとした場合。
- `FileNotFoundError`: 存在しないファイルにアクセスした場合。
- `IndexError`: シーケンスのインデックスが範囲外の場合。

### `ValueError` の例
不適切な値（例: 文字列）を型変換関数に渡した場合に発生する。

```python
# 不正な値を int() に渡す
x = int("abc")
```

### `TypeError` の例
不適切なデータ型同士の演算や操作が行われた場合に発生する。

```python
# 異なる型同士を加算しようとする
result = "hello" + 5
```

### `ZeroDivisionError` の例
数値を0で割ろうとした場合に発生する。

```python
# 0で除算を試みる
result = 10 / 0
```

### `FileNotFoundError` の例
存在しないファイルやパスにアクセスした場合に発生する。

```python
# 存在しないファイルを開く
with open("nonexistent_file.txt", "r") as file:
    content = file.read()
```

### `IndexError` の例
シーケンス（リスト、タプルなど）の範囲外のインデックスにアクセスした場合に発生する。

```python
# リストの範囲外インデックスにアクセス
numbers = [1, 2, 3]
print(numbers[5])
```
