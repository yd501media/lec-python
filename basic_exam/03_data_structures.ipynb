# **Data Structures: List, Tuple, Dictionary, and Set**
Pythonには効率的にデータを扱うためのデータ構造が用意されています。
- **リスト**: 順序付きで変更可能なコレクション。
- **タプル**: 順序付きで変更不可能なコレクション。
- **辞書**: キーと値のペアでデータを管理するコレクション。
- **セット**: ユニークな要素のみを持つ順序のないコレクション。

---
# リスト
リストは、複数の要素を格納できるデータ構造で、順序付きかつ変更可能です。  
要素には異なる型のデータも混在させることができます。

### 単純なリストの作成
リストは、複数の要素を1つのデータ構造にまとめて格納することができます。  
要素は順序を持ち、インデックスを使用してアクセス可能です。

```python
# 数値を要素として格納したリスト
squares = [1, 4, 9, 16, 25]
print(squares)
```

```python
# 文字列を要素として格納したリスト
fruits = ["apple", "banana", "cherry", "orange"]
print(fruits)
```

### 異なる型のデータを含むリスト
リストでは、文字列、数値、ブール値など、異なる型のデータを混在させて格納することができます。  
Pythonのリストの柔軟性を示す特徴の1つです。

```python
# 異なる型のデータを格納したリスト
mixed_list = ["Alice", 25, True, 3.14]
print(mixed_list)
```

### 空のリストを作成
要素が1つも含まれていないタプルを作成する場合、角括弧 `[]` もしくは `list()` を使用します。  
これはリストを初期化する際や、特定のケースで使用されます。

```python
empty_list = []
print(f"{empty_list}, {type(empty_list)=}")
```

```python
empty_list = list()
print(f"{empty_list}, {type(empty_list)=}")
```

空のリストは、例えば `for` 文の結果を格納するための入れ物として活用できます。

```python
# 空のリストを作成
squared_numbers = []

# for文を使用して結果をリストに追加
for number in range(1, 6):  # 1から5までの数を処理
    squared_numbers.append(number ** 2)

# 結果を表示
print(squared_numbers)
```

### リストの要素にアクセス
リストの要素はインデックスを使用してアクセスできます。

```python
# リストの作成
fruits = ["apple", "banana", "cherry"]

# インデックスを使用してアクセス
print(fruits[0])  # 結果: apple
print(fruits[-1])  # 結果: cherry
```

リストはミュータブル（変更可能）です。

```python
print(fruits)

fruits[1] = "lemmon"
print(fruits)
```

# タプル
タプルは、複数の要素を格納できるデータ構造で、順序付きですが変更不可能です。  
イミュータブルな性質を持つため、安全にデータを保持できます。

### 単純なタプルの作成
タプルは複数の要素を格納できるデータ構造で、リストと異なり変更不可能（イミュータブル）です。  
タプルは丸括弧 () を使用して作成します。

```python
# タプルの作成
fruits = ("apple", "banana", "cherry")
print(fruits)
```

### 異なる型のデータを含むタプル
タプルは異なる型のデータを混在させることができます。  
文字列、数値、ブール値、浮動小数点数などの異なるデータ型を1つのタプルに格納可能です。

```python
# 異なる型のデータを持つタプル
person = ("Alice", 25, True, 3.14)
print(person)
```

### 空のタプルを作成
要素が1つも含まれていないタプルを作成する場合、空の括弧 `()` もしくは `tuple()` を使用します。  

```python
# 空のタプル
empty_tuple = ()
print(f"{empty_tuple}, {type(empty_tuple)=}")
```

```python
# 空のタプル
empty_tuple = tuple()
print(f"{empty_tuple}, {type(empty_tuple)=}")
```

### タプルの要素にアクセス
タプルの要素もリストと同様にインデックスを使用してアクセスします。

```python
# タプルの作成
colors = ("red", "green", "blue")

# インデックスを使用してアクセス
print(colors[1])  # 結果: green
```

ただし、タプルはイミュータブル（変更不可）です。

```python
colors[1] = "black"
```

# 辞書
辞書は、キーと値のペアでデータを管理するデータ構造です。  
キーはユニークでなければならず、値にはどのような型でも使用できます。

### 基本的な辞書の作成
辞書はキーと値のペアでデータを管理します。キーはユニークである必要があります。  

```python
# 基本的な辞書の作成
person = {
    "name": "Alice",
    "age": 30,
    "is_employed": True
}
print(person)
```

以下のように `dict()` を使用して辞書を作成することもできます。  
この方法では、キーは文字列で指定します。

```python
# dict() を使用して辞書を作成
person = dict(name="Charlie", age=35, is_employed=False)
print(person)
```

### 異なる型のキーと値を持つ辞書
辞書はキーに異なる型のデータを使用できます。

```python
# 異なる型のキーと値を持つ辞書
mixed_dict = {
    "name": "Diana",
    42: "Answer",
    (1, 2): [3, 4, 5]  # タプルをキーに使用
}
print(mixed_dict)
```

### ネストされた辞書
辞書の値として別の辞書を格納することができます。

```python
# ネストされた辞書
employee = {
    "name": "Eve",
    "details": {
        "position": "Developer",
        "skills": ["Python", "JavaScript"]
    }
}
display(employee)
```

### 空の辞書を作成
要素が1つも含まれていない辞書を作成する場合、空の波括弧 {} もしくは dict() を使用します。  
これは辞書を初期化する際や、後からキーと値を追加する場合に使用されます。

```python
# 空の辞書
empty_dict = {}
print(f"{empty_dict}, {type(empty_dict)=}")
```

```python
# 空の辞書
empty_dict = dict()
print(f"{empty_dict}, {type(empty_dict)=}")
```

空の辞書を初期化し、後からキーと値を追加します。

```python
# 空の辞書を作成
data = {}

# キーと値を追加
data["name"] = "Bob"
data["age"] = 25

print(data)
```

### 辞書の要素にアクセス
辞書の要素には、キーを使用してアクセスします。

```python
# 辞書の作成
person = {"name": "Alice", "age": 25, "city": "New York"}

# キーを使用してアクセス
print(person["name"])  # 結果: Alice
print(person["age"])   # 結果: 25
```

例えば以下のようにfor文とメソッドを使用して要素にアクセスすることができます。

```python
for key in person.keys():
    print(key)
```

```python
for value in person.values():
    print(value)
```

```python
for key, value in person.items():
    print(f"{key=}, {value=}")
```

# セット
セットは、ユニークな要素を持つデータ構造で、順序が保証されません。  
重複を排除し、集合演算（和、積、差など）に便利です。

### 基本的なセットの作成
セットは、重複を許さないコレクションで、波括弧 {} を使用して作成します。

```python
# 基本的なセットの作成
fruits = {"apple", "banana", "cherry"}
print(fruits)
```

### 空のセットを作成
空のセットを作成するには、set() を使用します。{} は空の辞書として解釈されるため注意が必要です。

```python
# 空のセットを作成
empty_set = set()
print(f"{empty_set}, {type(empty_set)=}")
```

### 重複を排除する
セットを使用すると、リストやタプル内の重複を簡単に排除できます。

```python
# リストからセットを作成して重複を排除
numbers = [1, 2, 2, 3, 4, 4, 5]
print(numbers)
unique_numbers = set(numbers)
print(unique_numbers)
```

### 要素の追加と削除
セットは変更可能なデータ構造で、要素の追加や削除が可能です。

```python
fruits = {"apple", "banana", "cherry"}
print(fruits)

# 要素の追加
fruits.add("orange")
print(fruits)

# 要素の削除
fruits.remove("banana")
print(fruits)
```

### セットの要素にアクセス
セットは順序を持たないため、インデックスで直接アクセスすることはできません。  
ループや要素の存在確認でアクセスします。

```python
# セットの作成
numbers = {1, 2, 3, 4, 5}

# 要素の存在確認
print(3 in numbers)  # 結果: True
print(6 in numbers)  # 結果: False

# ループでアクセス（順序は保証されない）
for number in numbers:
    print(number)
```

# リスト操作
Pythonのリストでは、要素を追加・削除したり、部分的なデータを取り出すスライス操作、  
効率的なリスト生成を行うリスト内包表記など、さまざまな操作が可能です。

### 要素の追加
リストに新しい要素を追加するには、`append()` メソッドや `insert()` メソッドを使用します。
  
`append()`: リストの末尾に要素を追加します。

```python
# append() を使用して末尾に要素を追加
fruits = ["apple", "banana"]
fruits.append("cherry")
print(fruits)
```

`insert(index, value)`: 指定した位置に要素を挿入します。

```python
# insert() を使用して指定した位置に要素を挿入
fruits.insert(1, "orange")
print(fruits)
```

### 要素の削除
リストから要素を削除するには、`remove()` や `pop()` を使用します。
  
`remove(value)`: 指定した値を削除します。

```python
# remove() を使用して指定した値を削除
fruits = ["apple", "banana", "cherry"]
fruits.remove("banana")
print(fruits)
```

`pop(index)`: 指定したインデックスの要素を取り出し、リストから削除します。

```python
# pop() を使用して指定したインデックスの要素を削除
removed_item = fruits.pop(0)  # インデックス0の要素を削除
print(removed_item)  # 結果: apple
print(fruits)        # 結果: ['cherry']
```

### スライス
リストの一部を取得するにはスライスを使用します。
  
`list[start:end]`: 開始インデックスから終了インデックスの直前までを取得することができます。

```python
# スライスを使用して部分リストを取得
numbers = [1, 2, 3, 4, 5]
print(numbers[1:4])
```

スライスを使ってリストの一部を置き換えることも可能です。

```python
# スライスを使用して部分リストを置き換え
numbers[1:3] = [20, 30]
print(numbers)
```

### リスト内包表記
リスト内包表記は、1行で簡潔に新しいリストを作成する方法です。
  
条件を指定したり、要素を加工した結果を新しいリストに格納します。

```python
# 1〜10の数値から偶数のみを含むリストを作成
even_numbers = [x for x in range(1, 11) if x % 2 == 0]
print(even_numbers)
```

```python
# 文字列リストをすべて大文字に変換
words = ["hello", "world"]
uppercase_words = [word.upper() for word in words]
print(uppercase_words)
```

# タプルの特性
タプルは、Pythonのイミュータブル（変更不可能）なデータ構造です。  
この性質により、一度作成されたタプルの要素を変更することはできません。  
そのため、変更を防ぐ必要があるデータを安全に保持する場合や、関数の戻り値として複数の値を返す際に便利です。

## 特徴と用途
1. **イミュータブルな性質**:
    - タプルは作成後に要素の変更や追加・削除ができません。
    - データの安全性を確保する必要がある場合に適しています。
2. **用途（関数の戻り値として使用）**:
    - 関数が複数の値を返す際、タプルを使用して1つの戻り値としてまとめることができます。

### 1. イミュータブルな性質
タプルはイミュータブル（変更不可能）なデータ構造で、一度作成された要素は変更、追加、削除することができません。
この特性により、データの安全性を確保できるため、意図しない変更を防ぎたい場合に適しています。

```python
# タプルの作成
colors = ("red", "green", "blue")

# 要素へのアクセスは可能
print(colors[0])  # 結果: red

# 要素の変更は不可
colors[0] = "yellow"  # エラー: TypeError: 'tuple' object does not support item assignment
```

### 2. 関数の戻り値として使用
タプルは、関数が複数の値を返す際に非常に便利です。  
タプルを利用することで、複数の値を1つの戻り値として簡潔に返すことができ、コードの可読性が向上します。

```python
# 関数が複数の値を返す例
def calculate_circle_properties(radius):
    import math
    circumference = 2 * math.pi * radius  # 円周
    area = math.pi * (radius ** 2)        # 面積
    return circumference, area  # タプルとして返す

# 関数の戻り値を受け取る
result = calculate_circle_properties(5)
print(f"{result=}, {type(result)=}")
```

複数の値を1つの戻り値として簡潔に返すことができますが、展開(アンパック)して個別に受け取ることもできます。

```python
# タプルの要素を個別に展開して受け取る
circumference, area = calculate_circle_properties(5)
print(f"Circumference: {circumference}, Area: {area}")
```

# 辞書操作
辞書はキーと値のペアでデータを管理する強力なデータ構造です。  
以下では、辞書操作の主要な機能を紹介します。

### キーの存在確認
辞書内に特定のキーが存在するかどうかを確認するには、in キーワードを使用します。

```python
# 辞書の作成
person = {"name": "Alice", "age": 30, "city": "New York"}

# キーの存在確認
print("name" in person)  # 結果: True
print("salary" in person)  # 結果: False
```

### 値の更新
既存のキーの値を変更する場合や、新しいキーと値を追加する場合には、キーを指定して値を代入します。

```python
# 辞書の作成
person = {"name": "Alice", "age": 30}

# 値の更新
person["age"] = 31
print(person)

# 新しいキーと値を追加
person["city"] = "New York"
print(person)
```

### 辞書内包表記
辞書内包表記を使うと、1行で効率的に新しい辞書を作成できます。  
条件を指定したり、キーや値を加工して辞書を生成する際に便利です。

```python
# 辞書内包表記の例
numbers = [1, 2, 3, 4, 5]
squared_dict = {num: num**2 for num in numbers}
print(squared_dict)

# 条件付き辞書内包表記
filtered_dict = {num: num**2 for num in numbers if num % 2 == 0}
print(filtered_dict)
```

# セット操作
セットはユニークな要素を持つデータ構造で、集合演算を効率的に扱うことができます。  
以下では、セットを使用した 和集合、積集合、差集合 の操作について説明します。

### 和集合
和集合は、2つのセットに含まれる全ての要素を1つにまとめた集合です。  
`|` 演算子または `union()` メソッドを使用して計算します。

```python
# 2つのセットを作成
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# 和集合の計算
union_set = set_a | set_b
print(union_set)

# メソッドを使用
union_set_method = set_a.union(set_b)
print(union_set_method)
```

### 積集合
積集合は、2つのセットの両方に共通して含まれる要素の集合です。  
`&` 演算子または `intersection()` メソッドを使用して計算します。

```python
# 2つのセットを作成
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# 積集合の計算
intersection_set = set_a & set_b
print(intersection_set)

# メソッドを使用
intersection_set_method = set_a.intersection(set_b)
print(intersection_set_method)
```

### 差集合
差集合は、1つ目のセットに含まれ、2つ目のセットに含まれない要素の集合です。  
`-` 演算子または `difference()` メソッドを使用して計算します。

```python
# 2つのセットを作成
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# 差集合の計算
difference_set = set_a - set_b
print(difference_set)

# メソッドを使用
difference_set_method = set_a.difference(set_b)
print(difference_set_method)
```
