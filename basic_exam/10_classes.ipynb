# **Classes**
Pythonのクラスは、オブジェクト指向プログラミングの基本です。  
クラスはデータ（属性）と機能（メソッド）をまとめた設計図で、インスタンスを通じて操作します。


---
# クラス定義の構文
Pythonでは、class キーワードを使用してクラスを定義します。  
クラス内には、属性（変数）やメソッド（関数）を定義できます。

```python
class SampleClass:
    """サンプルクラス"""

    def __init__(self, name: str, value: int):
        """
        コンストラクタ
        :param name: 名前 (文字列)
        :param value: 値 (整数)
        """
        self.name = name
        self.value = value

    def display_info(self):
        """
        オブジェクトの情報を表示するメソッド
        """
        print(f"Name: {self.name}, Value: {self.value}")

    def increment_value(self, increment: int):
        """
        値を増加させるメソッド
        :param increment: 増加量
        """
        self.value += increment

# インスタンスオブジェクト生成
sample = SampleClass("SampleObject", 10)
# インスタンス変数を表示
sample.display_info()
# 値を増加させるメソッドを呼び出し
sample.increment_value(5)
# インスタンス変数を再表示
sample.display_info()
```


---
# クラスオブジェクト
クラス自体もオブジェクトであり、`type` クラスのインスタンスとして振る舞います。  
クラスには属性やメソッドが定義され、これを基にインスタンスが生成されます。

```python
print(SampleClass)
```


---
# インスタンスオブジェクト
クラスを元に生成される具体的なオブジェクト。  
各インスタンスは独立した属性を持つことができます。

```python
# インスタンスオブジェクトを生成
obj1 = SampleClass("Alice", 5)
obj2 = SampleClass("Bob", 10)

print(f"{type(obj1)=}")
print(f"{type(obj2)=}")
```


---
# メソッドオブジェクト
クラス内で定義された関数（メソッド）は、インスタンスやクラスに関連付けられたオブジェクトとして振る舞います。

```python
print(f"{type(obj1.display_info)=}")
print(f"{type(obj2.display_info)=}")
```

以下のように指定することで、メソッドを呼び出すことができます。

```python
# インスタンス変数を表示するメソッドを呼び出し
obj1.display_info()
obj2.display_info()
```

```python
# インスタンス変数を増加させるメソッドを呼び出し
obj1.increment_value(5)
obj2.increment_value(10)

# インスタンス変数を表示するメソッドを再度呼び出し
obj1.display_info()
obj2.display_info()
```


---
# クラス変数とインスタンス変数
- **クラス変数**: クラス全体で共有される変数。
- **インスタンス変数**: 各インスタンスが個別に持つ変数。

```python
class SampleClass:
    class_var = "shared"  # クラス変数

    def __init__(self, name):
        self.name = name  # インスタンス変数

# インスタンス生成
obj1 = SampleClass("Alice")
obj2 = SampleClass("Bob")

# クラス変数を表示
print(f"{obj1.class_var=}")
print(f"{obj2.class_var=}")

# インスタンスオブジェクトが異なっていても、クラス変数が同一であることを確認
print(f"{(obj1.class_var==obj2.class_var)=}", end="\n\n")

# インスタンス変数を表示
print(f"{obj1.name=}")
print(f"{obj2.name=}")
```

# 継承
親クラスの機能を子クラスが引き継ぐことができます。  

以下の例では親クラスである `BaseClass` を定義し、その中でメソッドとして `func_a` と `func_b` を定義しています。  
その上で、子クラス `ChildClass` を定義する際に `BaseClass` を継承しつつ、クラスメソッドとして `func_c` を定義しています。

```python
# 親クラス
class BaseClass():
    def __init__(self, instance_name: str):
        self.instance_name = instance_name

    def func_a(self):
        print(f"This instance is generated from {self.instance_name}.\nThis method is `func_a` defined in BaseClass.\n")

    def func_b(self):
        print(f"This instance is generated from {self.instance_name}.\nThis method is `func_b` defined in BaseClass.\n")

# 子クラス, 親クラスを継承する
class ChildClass(BaseClass):
    def __init__(self, instance_name: str):
        self.instance_name = instance_name

    def func_c(self):
        print(f"This instance is generated from {self.instance_name}.\nThis method is `func_c` defined in ChildClass.\n")
```

以下では、親クラスからインスタンスオブジェクトを生成して、クラスメソッドを呼び出しています。

```python
# 親クラスからインスタンスを生成
cls_base = BaseClass("BaseClass")
# 親クラスで定義されているメソッドを呼び出し
cls_base.func_a()
cls_base.func_b()
```

以下では、子クラスからインスタンスオブジェクトを生成して、クラスメソッドを呼び出しています。  
その後、継承した親クラスで定義されているクラスメソッドを呼び出しています。

```python
# 子クラスからインスタンスを生成
cls_child = ChildClass("ChildClass")
# 子クラスで定義されているメソッドを呼び出し
cls_child.func_c()
# 子クラスで継承した親クラスで定義されているメソッドを呼び出し
cls_child.func_a()
cls_child.func_b()
```

子クラスは親クラスのメソッドや属性をオーバーライドすることも可能です。

```python
# 親クラス
class Animal:
    def speak(self):
        return "I am an animal"

# 子クラスで親クラスのspeak()メソッドをoverride
class Dog(Animal):
    def speak(self):
        return "Woof!"
    
# 親クラスの speak() メソッドを実行
animal = Animal()
print(f"{animal.speak()=}")

# 子クラス
dog = Dog()
print(f"{dog.speak()=}")
```

# 多重継承
1つのクラスが複数の親クラスから継承することが可能です。

以下の例では、親クラスとして `ClassA` と `ClassB` を定義しています。  
その上で、子クラス `ClassC` を定義する際に `ClassA` と `ClassB` を継承しています。

```python
# 親クラスA
class ClassA:
    def feature_a(self):
        return "Feature A"

# 親クラスB
class ClassB:
    def feature_b(self):
        return "Feature B"

# 子クラス
class ClassC(ClassA, ClassB):
    pass
```

子クラスである `ClassC` ではメソッドを定義していませんが、親クラスからメソッドを継承しているため、  
`feature_a` と `feature_b` を呼び出すことができます。

```python
obj = ClassC()
print(f"{obj.feature_a()=}")
print(f"{obj.feature_b()=}")
```
