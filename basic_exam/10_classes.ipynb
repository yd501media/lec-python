# **Classes**
Pythonのクラスは、オブジェクト指向プログラミングの基本です。  
クラスはデータ（属性）と機能（メソッド）をまとめた設計図で、インスタンスを通じて操作します。

---
- **クラス**
    - データと機能をまとめた設計図。
- **メソッドとプロパティ**
    - クラス内で定義された関数と属性。
- **クラスオブジェクト**
    - クラス自体を表すオブジェクト。
- **インスタンスオブジェクト**
    - クラスから生成された具体的なオブジェクト。
- **メソッドオブジェクト**
    - クラスやインスタンスに紐づいた関数。
- **クラス変数とインスタンス変数**
    - クラス全体共有の変数とインスタンス専用の変数。
- **継承**
    - 親クラスの機能を引き継ぐ仕組み。
- **多重継承**
    - 複数の親クラスから機能を継承。
- **抽象基底クラス**
    - 子クラスで実装を強制するクラス。
