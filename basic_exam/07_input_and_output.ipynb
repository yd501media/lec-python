# **Input and Output Operations**

これらの操作は、プログラムの基本的な構成要素であり、ユーザーとのインタラクションやファイル操作、外部データの読み書きに役立ちます。

# 文字列
Pythonでは、文字列のフォーマットを使用して動的にデータを組み込むことができます。  
`f文字列`、`str.format()`、`%演算子` などの方法を活用することで、柔軟かつ簡潔に文字列を操作できます。 

### f文字列
Python 3.6以降で導入された `f文字列`は、文字列内に直接変数や式を埋め込むことができ、可読性が高いフォーマット方法です。  
文字列の前に `f` を付けて使用します。

```python
name = "Alice"
age = 25

# f文字列を使用
message = f"My name is {name} and I am {age} years old."
print(message)

# 式の埋め込み
print(f"Next year, I will be {age + 1} years old.")
```

### str.format()
`str.format()` メソッドは、文字列内のプレースホルダー `{}` に値を埋め込むための方法です。  
プレースホルダーにインデックスやキーワードを指定することで、柔軟なフォーマットが可能です。

```python
name = "Bob"
age = 30

# 基本的な使用例
message = "My name is {} and I am {} years old.".format(name, age)
print(message)

# インデックスを指定
message = "I am {1} years old, and my name is {0}.".format(name, age)
print(message)

# キーワードを使用
message = "My name is {name} and I am {age} years old.".format(name="Charlie", age=35)
print(message)
```

### %演算子
`%演算子` は、C言語風の文字列フォーマット方法です。  
`%` を使って値を埋め込みますが、現在では `f文字列` や `str.format()` が推奨されています。

```python
name = "Diana"
age = 28

# 基本的な使用例
message = "My name is %s and I am %d years old." % (name, age)
print(message)

# 浮動小数点のフォーマット
pi = 3.14159
print("The value of pi is approximately %.2f." % pi)
```

# 標準出力
`print()` 関数を使用して、コンソールにデータを表示します。  
基本的な表示から、複数の値をフォーマットして出力することが可能です。プログラムの結果を確認するために最も一般的に使用される手段です。

## `print()` を使用したデータの表示
Pythonでは、print() 関数を使用して標準出力（コンソール画面）にデータを表示することができます。  
単純な文字列や数値、複数の値、フォーマットされた出力を表示する際に使用します。

### 基本的な使い方
`print()` 関数を使用して文字列や数値をコンソールに表示します。  
単一のデータを出力する最も基本的な使い方です。

```python
# 文字列を表示
print("Hello, World!")  

# 数値を表示
print(42)  
```

### 複数の値を表示
`print()` に複数の値を渡すことで、デフォルトで空白で区切られた出力を生成します。  
これにより、簡単に複数のデータを連結して表示できます。

```python
# 複数の値を表示（デフォルトは空白区切り）
print("The answer is", 42)
```

### `sep` を使用
`sep` パラメータを指定することで、複数の値を区切る文字列をカスタマイズできます。  
デフォルトでは空白（`' '`）が使用されますが、コンマやハイフンなど任意の文字列に変更可能です。

```python
# 区切り文字を変更
print("apple", "banana", "cherry", sep=", ")
```

### `end` を使用
`end` パラメータを指定することで、出力の末尾に追加される文字列を変更できます。  
デフォルトでは改行（`\n`）ですが、これを空白や他の文字に変更して、連続する出力を1行にまとめることが可能です。

```python
# 改行をカスタマイズ
print("Hello", end=" ")
print("World!")
```

以下のように出力末尾の文字列を `\n\n` とした場合は、追加で改行を加えることができます。

```python
print("Hello, world!", end="\n\n")
print("This is Python.")
```

以下のように出力末尾の文字列を `""` とすることで、改行を削除することができます。

```python
import time
print("wait ", end="")
for n in range(10):
    print(".", end="")
    time.sleep(1)
else:
    print("\nDone!")
```

### フォーマットされた出力
文字列内に変数や式を埋め込む際に、`f文字列` を使用することで簡潔で可読性の高いフォーマットが可能です。  
これにより、動的なデータを含むメッセージを生成できます。

```python
# f文字列でのフォーマット
name = "Alice"
age = 25
print(f"My name is {name} and I am {age} years old.")
```

# 標準入力
`input()` 関数を使用して、ユーザーからのデータ入力を取得します。  
入力されたデータは文字列として処理されるため、必要に応じて型変換を行います。プログラムにインタラクティブな操作を追加する際に役立ちます。
### 基本的な使用例

```python
# ユーザーから名前を入力
# 入力例: YourName
name = input("What is your name? ")
print(f"Hello, {name}!")
```

### 型変換を伴う入力
`input()` から取得するデータは文字列型 (`str`) で返されるため、数値として使用する場合には型変換が必要です。

```python
# ユーザーから年齢を入力
# 入力例: 25
age = int(input("How old are you? "))
print(f"You are {age} years old.")
```

### 複数のデータを入力
複数の値を一度に入力し、分割して処理する場合には、`split()` を使用します。

```python
# 複数の数値を入力
# 入力例: 5 10 15
numbers = input("Enter three numbers separated by spaces: ").split()
# 数値として処理するために型変換
numbers = [int(num) for num in numbers]
print(f"Sum of numbers: {sum(numbers)}")
```

### デフォルト値を使用する
ユーザーが入力を省略した場合に、デフォルト値を使用する処理を追加します。

```python
# 名前を入力（省略時は "Guest"）
name = input("What is your name? (default: Guest) ") or "Guest"
print(f"Welcome, {name}!")
```

# ファイル操作
Pythonの組み込み関数 `open()` を使用して、テキストやバイナリファイルを読み書きできます。  
`read()` でファイルの内容を取得し、`write()` でデータを書き込むなど、基本的なファイル操作を簡単に実現できます。

## `open()`, `read()`, `write()` の基本操作

### ファイルの読み込み (`read()`)
`open()` 関数を使ってファイルを開き、`read()` を使用してファイル全体の内容を一括で読み取ります。  
`with` 文を使用することで、ファイルのクローズを自動的に処理し、リソースリークを防ぎます。

```python
# ファイルを読み込む
with open("./samples/example_input.txt", "r", encoding="utf-8") as file:
    content = file.read()
print(content)
```

### ファイルへの書き込み (`write()`)
`open()` 関数で `'w'` モードを指定すると、指定したファイルに書き込みが行われます。  
このモードでは、既存のファイルが上書きされ、新規ファイルは作成されます。

```python
# ファイルに書き込む
with open("./samples/example_output.txt", "w", encoding="utf-8") as file:
    file.write("Hello, World!\nThis is a sample file.")
```

### 追記モード (`append`)
`open()` 関数で `'a'` モードを指定すると、既存のファイルにデータを追記します。  
このモードでは、元のファイル内容は保持され、末尾に新しいデータが追加されます。

```python
# ファイル末尾に追記する
with open("./samples/example_output.txt", "a", encoding="utf-8") as file:
    file.write("\nAdding more text to the file.")
```

### ファイルを1行ずつ読み込み
`for` 文を使用して、ファイルを1行ずつ読み取ります。この方法は、大きなファイルを扱う際に便利で、メモリ効率が良いです。  
`strip()` メソッドを使用することで、行末の改行文字を取り除いて表示できます。

```python
# ファイルを1行ずつ読み込む
with open("./samples/example_input.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())  # 改行を削除して出力
```

# CSVやJSONの操作
Pythonでは、`csv` モジュールや `json` モジュールを使用して外部データファイルを扱うことができます。  
CSV形式の表データやJSON形式の構造化データを読み書きする際に役立ち、データ分析やWeb APIの利用に欠かせない機能です。

## CSV操作
`csv` モジュールを使用して、CSVファイル（カンマ区切りの表形式データ）の読み書きが可能です。

### CSVファイルの読み込み
`csv.reader` を使用して、CSVファイルを行ごとに読み込みます。  
ファイルは `open()` を使って開き、`reader` オブジェクトを反復処理して各行をリストとして取得します。

```python
import csv

# CSVファイルを読み込む
with open("./samples/data_input.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

### CSVファイルへの書き込み
`csv.writer` を使用してデータをCSVファイルに書き込みます。  
`writerows()` を使うと、リストのリストを一括で書き込むことができます。

```python
import csv

# データをCSVに書き込む
data = [
    ["Name", "Age", "City"],
    ["Alice", 25, "New York"],
    ["Bob", 30, "Los Angeles"]
]

with open("./samples/data_output.csv", "w", encoding="utf-8", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)
```

## JSON操作
`json` モジュールを使用して、JSON形式のデータ（階層構造を持つデータ）を扱うことができます。

### JSONファイルの読み込み
`json.load` を使用して、JSON形式のファイルをPythonの辞書やリストに変換して読み込みます。

```python
import json

# JSONファイルを読み込む
with open("./samples/data_input.json", "r", encoding="utf-8") as file:
    data = json.load(file)
    print(data)
```

### JSONファイルへの書き込み
`json.dump` を使用して、Pythonの辞書やリストをJSON形式でファイルに保存します。  
`indent` パラメータを指定すると、読みやすい整形されたフォーマットで保存できます。

```python
import json

# データをJSONに書き込む
data = {
    "Name": "Bob",
    "Age": 30,
    "City": "Los Angeles"
}

with open("./samples/data_output.json", "w", encoding="utf-8") as file:
    json.dump(data, file, indent=4)
```
