# **Input and Output Operations**
Pythonでは、`input()`を使ってユーザーからデータを取得し、`print()`を使って出力します。  
また、ファイル操作では`open()`を使用してデータの読み書きが可能です。

---
- **文字列**
    - フォーマット
- **標準出力**
    - `print()` を使用したデータの表示
- **標準入力**
    - `input()` を使用したデータの取得
- **ファイル操作**
    - `open()`, `read()`, `write()` の基本操作
- **CSVやJSONの操作**
    - 外部データファイルの入出力
