# **Virtual Environments**
仮想環境は、Pythonプロジェクトごとに異なる依存関係を分離して管理するための仕組みです。  
これにより、プロジェクト間のパッケージの競合を防ぎます。  
Pythonの標準ライブラリである `venv` を使用して簡単に作成・有効化・無効化できます。


---
# 仮想環境の作成
以下のコマンドで仮想環境を作成します。

```bash
python -m venv .venv
```

`.venv` は仮想環境の名前です。この名前は任意に変更可能です。  
このコマンドを実行すると、`.venv` ディレクトリが作成され、その中に仮想環境が構築されます。


---
# 仮想環境の有効化
作成した仮想環境を有効化するには、OSに応じたコマンドを実行します。

## Windows:
```bash
.venv\Scripts\activate
```
## macOS/Linux:
```bash
. .venv/bin/activate
```

## 有効化の確認:
仮想環境が有効になると、ターミナルの左側に仮想環境名 `(.venv)` が表示されます。
```bash
(.venv) $
```


---
# 仮想環境の無効化
仮想環境を無効化するには以下を実行します。
```bash
deactivate
```
