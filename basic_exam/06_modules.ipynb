# **Modules**
モジュールはPythonでコードを再利用可能にする単位です。  
`math`や`random`などの組み込みモジュールや、`numpy`、`pandas`などの外部ライブラリを使用して、追加の機能を簡単に利用できます。


---
# モジュールのインポート方法
Pythonでは、外部ライブラリや標準ライブラリを活用するためにモジュールをインポートします。
- **`import`**: モジュール全体をインポートします。
- **`from ... import`**: モジュール内の特定の関数やクラスをインポートします。
- **`import ... as ...`**: モジュール全体をインポートし、別名を付けて使用します。

## **import ...**
Pythonでは、`import` を使用することでモジュール全体をプログラム内に取り込み、  
そのモジュールが提供するすべての関数やクラスを利用できるようになります。
  
以下は `math` モジュールを読み込み、その中から `sqrt()` を呼び出して使用する例です。

````python
# import を使用してモジュール全体をインポート
import math
print(math.sqrt(16))
````

例えば、以下のように `関数` ではなく `定数` を呼び出すことも可能です。

````python
print(f"{math.pi=}")
print(f"{math.e=}")
````

## **from ... import ...**
`from ... import ...` を使用することでモジュール全体をインポートせず、  
特定の関数やクラス、定数だけをプログラム内に取り込むことができます。  
以下の例の場合、`math.` を省略することが可能です。

````python
# from ... import を使用して特定の関数をインポート
from math import sqrt
print(sqrt(25))
````

以下のように、複数の特定の関数などを `import` することも可能です。

````python
from math import sin, cos, pi, e

print(f"{sin(0)=}")
print(f"{cos(0)=}")
print(f"{pi=}")
print(f"{e=}")
````

## **import ... as ...**
`import ... as ...` を使用すると、モジュール全体をインポートした際に、短縮名やカスタム名を付けて利用することができます。  
コードの簡潔化や可読性の向上、名前の衝突を回避する際に役立ちます。
  
以降で説明する `numpy` や `pandas` などでよく使用される構文です。
```python
import numpy as np
import pandas as pd
```

# 外部モジュールのインストール
Pythonでは、`pip` を使用して外部モジュールをインストールできます。  
インストール後、通常のモジュールと同様に `import` して使用します。  
例として、`numpy` と `pandas` モジュールをインストールします。
  
以下のように `コマンドプロンプト` や `terminal` から `pip` コマンドを使用してモジュールをインストールできます。
```bash
# Windowsの場合
pip install numpy
pip install pandas

# Mac, Linuxの場合
pip3 install numpy
pip3 install pandas
```
  
または以下のようにノートブックファイル内から `pip` コマンドを使用してインストールすることもできます。

````python
# Windowsの場合
!pip install numpy pandas

# Mac, Linuxの場合
!pip3 install numpy pandas
````

# numpy
`numpy` は数値計算を効率的に行うためのライブラリで、多次元配列や数学関数を提供します。

````python
import numpy as np

# 配列の作成
array = np.array([1, 2, 3, 4])
print(array)  # 結果: [1 2 3 4]

# 配列の演算
print(array * 2)  # 結果: [2 4 6 8]
````

# pandas
`pandas` はデータ分析に特化したライブラリで、データフレームやシリーズを扱うための便利な機能を提供します。

````python
import pandas as pd

# データフレームの作成
data = {"Name": ["Alice", "Bob", "Charlie"], "Age": [25, 30, 35]}
df = pd.DataFrame(data)

display(df)
````
