# **Modules**
モジュールはPythonでコードを再利用可能にする単位です。  
`math`や`random`などの組み込みモジュールや、`numpy`、`pandas`などの外部ライブラリを使用して、追加の機能を簡単に利用できます。

---
- **モジュールのインポート方法**
    - `import`, `from ... import`
- **その他の基本モジュール**
    - `math`, `random`, `os`
- **外部モジュールのインストール**
    - `pip install`
