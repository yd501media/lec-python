# **Functions**
関数は、再利用可能なコードブロックを定義するために使用します。  
引数や返り値を使うことで、柔軟に処理をカスタマイズできます。  
`lambda`式を使うことで、簡潔な無名関数を定義することも可能です。


---
# 関数の定義
関数は `def` キーワードを使用して定義します。  
再利用可能な処理をまとめ、コードの効率化と可読性を向上させます。

```python
# 関数の定義
def greet(name):
    return f"Hello, {name}!"

# 関数の呼び出し
message = greet("Alice")
print(message)
```


---
# 引数と返り値
関数は引数を受け取り、処理結果を返り値として返すことができます。

```python
def add(a, b):
    """2つの数値を加算して返す"""
    return a + b

print(add(3, 5))
```

### 引数のデフォルト値
引数にデフォルト値を設定すると、引数を省略した場合にそのデフォルト値が使用されます。

```python
def greet(name="Guest"):
    """デフォルト値を持つ引数の例"""
    return f"Hello, {name}!"

print(f"{greet()=}")
print(f"{greet("Alice")=}")
```

### キーワード引数
関数を呼び出す際に、引数名を指定して値を渡す方法です。順序に依存せず、可読性が向上します。

```python
def introduce(name, age):
    """キーワード引数の例"""
    return f"{name} is {age} years old."

print(introduce(age=25, name="Bob"))
```

### 特殊引数
特殊引数を使用して、柔軟に引数を受け取ることができます。

- `*args`: 任意の位置引数をタプルとして受け取ります。
- `**kwargs`: 任意のキーワード引数を辞書として受け取ります。

```python
def demo_args(*args, **kwargs):
    """特殊引数の例"""
    print("args:", args)
    print("kwargs:", kwargs)

demo_args(1, 2, 3, name="Alice", age=30)
```

### 引数リストのアンパック
リストや辞書を引数として渡す際に、アンパック（展開）して渡すことができます。

```python
def add(a, b, c):
    """引数リストのアンパックの例"""
    return a + b + c

nums = [1, 2, 3]
print(f"{add(*nums)=}")

kwargs = {"a": 4, "b": 5, "c": 6}
print(f"{add(**kwargs)=}")
```

# 無名関数（lambda）
`lambda` を使って簡易的な関数を定義できます。短い処理に適しています。

```python
square = lambda x: x ** 2
print(f"{square(4)=}")

numbers = [1, 2, 3, 4]
doubled = list(map(lambda x: x * 2, numbers))
print(f"{doubled=}")
```

# docstring
`docstring` は関数やクラスの説明を記述するための文字列です。  
関数の直後に記述し、`help()` 関数で内容を確認できます。

```python
def greet(name):
    """名前を受け取って挨拶を返す関数"""
    return f"Hello, {name}!"

help(greet)
```

# 関数アノテーション (注釈)
関数アノテーションは、関数の引数や返り値にデータ型を明示的に指定するための記法です。  
Pythonの動的型付けには影響を与えませんが、コードの可読性を向上させ、  
静的解析ツールや型チェックツール（例: `mypy`）と併用することで、バグの早期発見に役立ちます。
  
- **引数のアノテーション**: 各引数の後に `:` を使用して型を指定します。
- **返り値のアノテーション**: 関数定義の末尾に `->` を使用して型を指定します。

### 基本的なアノテーション
以下の例では、`a: int` と `b: int` という記述で、引数 `a` と `b` が整数型であることを指定しています。  
また、`-> int` として、関数の返り値が整数型であることを指定しています。

```python
def add(a: int, b: int) -> int:
    """引数と返り値に型アノテーションを追加"""
    return a + b

print(add.__annotations__)
print(f"{add(3, 5)=}")
```

### 文字列のアノテーション
以下の例では、`name: str` という記述で、引数 `name` が文字列型であることを指定しています。  
また、`-> str` として、関数の返り値が文字列型であることを指定しています。

```python
def greet(name: str) -> str:
    """名前を受け取って挨拶を返す"""
    return f"Hello, {name}!"

print(greet.__annotations__)
print(f"{greet("Alice")=}")
```

### リストや辞書型の例
以下の例では、`items: list[int | float]` という記述で、  
引数 `items` が、要素が `int` もしくは `float` である `list` であることを指定しています。  
また、`-> dict` として、関数の返り値が辞書型であることを指定しています。

```python
def process_items(items: list[int | float]) -> dict:
    """リストを受け取り、その長さと合計を辞書で返す"""
    return {"count": len(items), "sum": sum(items)}

print(process_items.__annotations__)
print(f"{process_items([1, 2, 3])=}")
```

### None を返す関数
返り値がない関数は、アノテーションに `None` を使用します。  
返り値がない場合は、`-> None`を省略できます。

```python
def display_message(message: str) -> None:
    """メッセージを表示するだけで値は返さない"""
    print(f"Message: {message}")

print(display_message.__annotations__)
display_message("Hello!")
```

### デフォルト値のある引数
デフォルト値がある引数にもアノテーションを付けることができます。  
以下の例では`age: int = 30` と言う記述で、引数 `age` の型は整数を指定し、デフォルト値を `30` に設定しています。

```python
def describe_person(name: str, age: int = 30) -> str:
    """名前と年齢を返す"""
    return f"{name} is {age} years old."

print(describe_person.__annotations__)
print(f"{describe_person("Bob")=}")
print(f"{describe_person("Alice", 25)=}")
```
