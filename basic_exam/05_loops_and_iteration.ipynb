# **Loops and Iteration**
繰り返し処理は、`for`文や`while`文を使って実現できます。  
リストや辞書を反復処理することで、効率的にデータを操作することが可能です。


---
# for文
`for` 文は、リストやタプル、文字列などのシーケンスの各要素を順に処理するループ構文です。  

```python
# シーケンスを反復処理
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

また、`range()` 関数を使用することで指定した範囲の繰り返し処理も可能です。

```python
# range() を使用したループ
for i in range(5):
    print(i)
```


---
# while文
`while` 文は、指定した条件が真である間、繰り返し処理を実行します。  
条件が偽になるとループが終了します。

```python
# while文を使用した繰り返し
count = 0
while count < 5:
    print(count)
    count += 1
```


---
# break, continue, else, pass
- **`break`**: ループを途中で終了します。
- **`continue`**: その回の残りの処理をスキップして次の反復に進みます。
- **`else`**: ループが正常終了した場合に実行されます。
- **`pass`**: 何も処理をせずにスキップします。
  
以下は `for` 文を使用した繰り返し処理の中に `break`, `continue`, `else` を組み込んだ例です。  
このサンプルは5行目の `break` で終了(正常終了ではない)するため `else` 節は実行されません。

```python
# break, continue, else の例
for i in range(5):
    if i == 3:
        print(f"{i=}, ループを終了")
        break
    elif i == 1:
        print(f"{i=}, この回をスキップ")
        continue
    print(f"{i=}")
else:
    print("Loop completed")  # breakが呼ばれると実行されない
```

以下のサンプルは正常終了するため `else` 節が実行されます。

```python
for i in range(5):
    if i == 10:
        print(f"{i=}, ループを終了")
        break
    elif i == 1:
        print(f"{i=}, この回をスキップ")
        continue
    print(f"{i=}")
else:
    print("Loop completed")  # breakが呼ばれると実行されない
```

以下は `pass` を使用した例です。

```python
# passの使用例
for _ in range(3):
    pass  # 何も実行しない
```

`pass` を除外した場合は構文エラーとなります。

```python
for _ in range(3):
```

# enumerate, zip
- **`enumerate`**: シーケンスのインデックスと要素を同時に取得できます。
- **`zip`**: 複数のシーケンスを並行して処理します。  
`zip` に `strict=True` を指定すると、長さが異なるシーケンスを渡した場合にエラーを発生させます（Python 3.10以降）。

`enumerate()` は、シーケンス（リスト、タプル、文字列など）を反復処理する際に、  
各要素とそのインデックスを同時に取得できる便利な組み込み関数です。

```python
# enumerate の使用
fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits):
    print(f"Index: {index}, Fruit: {fruit}")
```

`zip()` は、複数のシーケンス（リスト、タプル、文字列など）の要素をペアとしてまとめ、新しいイテレータを生成する組み込み関数です。

```python
# zip の使用
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]
for name, age in zip(names, ages):
    print(f"{name} is {age} years old.")
```

`zip()` の `strict` オプションは、`Python 3.10` 以降で利用可能な機能で、入力シーケンスの長さが異なる場合に例外を発生させるオプションです。  
これにより、データ整合性を確保しやすくなります。

```python
# zip の strict オプション
names = ["Alice", "Bob"]
ages = [25, 30, 35]
for name, age in zip(names, ages, strict=True):  # エラー: 長さが異なる
    print(name, age)
```

# リスト内包表記のループ
リスト内包表記を使うと、簡潔に新しいリストを生成できます。  
条件を指定したり、要素を加工する場合にも利用されます。


以下は、要素が全て `0` のリストを生成する例です。

```python
zeros = [0 for n in range(10)]
print(zeros)
```

以下は、既存リストの要素を2乗して返す例です。

```python
# 基本的なリスト内包表記
numbers = [1, 2, 3, 4, 5]
squared = [n**2 for n in numbers]
print(squared)
```

以下は、リスト内包表記の中に条件文を組み込んだ例です。

```python
# 条件付きリスト内包表記
even_numbers = [n for n in numbers if n % 2 == 0]
print(even_numbers)
```
