# **Loops and Iteration**
繰り返し処理は、`for`文や`while`文を使って実現できます。  
リストや辞書を反復処理することで、効率的にデータを操作することが可能です。

---
- **for文**
    - シーケンスの各要素に対して繰り返し処理を実行
- **while文**
    - 条件による繰り返し
- **break, continue**
    - ループ制御
- **enumerate, zip**
    - 繰り返しでの便利な組み込み関数
    - zipのstrictオプション
- **リスト内包表記のループ**
    - 効率的なリスト生成
