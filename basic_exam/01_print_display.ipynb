# print() と display()
Pythonでは、結果や情報を出力する際に主に以下の2つの関数を使用します。

## print()
- テキスト形式でコンソールにデータを表示する関数。
- 主にデバッグやユーザーへのメッセージ表示に使用します。

```python
# 値を直接表示する
print(100)
print(3.14159)
print("Hello, world!")
```

```python
# 変数に値を代入して、それを表示する
num = 100
pi = 3.14159
greeting = "Hello, world!"
print(num)
print(pi)
print(greeting)
```

```python
# こんなこともできる
print(f"{num=}")
print(f"{pi=}")
print(f"{greeting=}")
```

## display()
- Pythonのインタラクティブな環境（Jupyter Notebookなど）で、データのフォーマットや構造を整えた状態で表示する関数。
- 特にデータフレームやHTMLコンテンツを見やすく表示する際に有用です。

```python
# dictの場合
data = {
    "name": "Alice",
    "age": 25,
    "city": "New York",
    "skills": ["Python", "Data Analysis", "Machine Learning"]
}

print("print()の場合")
print(data, end="\n\n===== ===== ===== ===== =====\n\n")

print("display()の場合")
display(data)
```

```python
# 以下のコードセル実行時、`import pandas as pd`でエラーが発生した場合は、
# 下行のコメントアウトを解除し`!pip3 install pandas`を実行する
# !pip3 install pandas
```

```python
# DataFrameの場合
import pandas as pd

df = pd.DataFrame(
    [[(row + col) for col in range(5)] for row in range(10)],
    columns=[f"col_{n}" for n in range(5)],
    index=[f"index_{n}" for n in range(10)],
)

print("print()の場合")
print(df, end="\n\n===== ===== ===== ===== =====\n\n")

print("display()の場合")
display(df)
```
