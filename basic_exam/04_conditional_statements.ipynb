# **Conditional Statements**
条件分岐は、`if`、`elif`、`else`を使ってプログラムの動作を条件によって分岐させます。  
`match-case`文を使用して複雑な条件分岐を簡潔に表現することもできます。

---
# if-else, elif
`if`、`elif`、`else` を使用して条件による分岐を行います。  
複数の条件を順次評価し、該当する条件のブロックを実行します。

```python
# 基本的な条件分岐
x = 10

if x > 20:
    print("x is greater than 20")
elif x > 10:
    print("x is greater than 10 but less than or equal to 20")
else:
    print("x is 10 or less")
```

---
# ネストされた条件分岐
`if` 文の中にさらに `if` 文を記述することで、より複雑な条件を評価できます。  
ただし、ネストが深くなると可読性が低下するため注意が必要です。

```python
# 入れ子構造の条件分岐
x = 15

if x > 10:
    if x < 20:
        print("x is between 10 and 20")
    else:
        print("x is 20 or more")
else:
    print("x is 10 or less")
```

---
# 三項演算子
条件分岐を1行で記述できる構文です。  
簡潔に書きたい場合に便利ですが、複雑なロジックには向きません。

```python
# 三項演算子を使用した条件分岐
x = 5
result = "x is even" if x % 2 == 0 else "x is odd"
print(result)
```

---
# match-case
Python 3.10から導入された構文で、複数の条件を簡潔に記述できます。  
従来の `if-elif-else` の代替として使用可能で、値のパターンマッチングも行えます。

```python
# match-caseを使用した条件分岐
command = "start"

match command:
    case "start":
        print("Starting the system...")
    case "stop":
        print("Stopping the system...")
    case "restart":
        print("Restarting the system...")
    case _:
        print("Unknown command")
```

## Conditional Statements
- **pass**
    - 何も処理をしない場合に使用する構文。
- **is**
    - オブジェクト同士が同一かを判定する演算子。
- **in**
    - 要素がシーケンスに含まれるかを確認する演算子。
- **not**
    - 論理値を反転させるブール演算子。


---
# pass
`pass` は条件分岐の中で何も処理を行わない場合に使用できます。  
処理が未定の場合や、条件を満たしても実行すべき操作がない場合に役立ちます。

```python
x = 10

if x > 5:
    pass  # 条件を満たしても何もしない
else:
    print("x is less than or equal to 5")
```


---
# not
`not` 演算子を使うと、条件の結果を反転させることができます。  
条件分岐で、特定の条件を満たさない場合の処理に役立ちます。

```python
x = 10

if not x < 5:
    print("x is not less than 5")
else:
    print("x is less than 5")
```


---
# is
`is` 演算子を使うと、2つのオブジェクトが同一かどうかを判定できます。  
条件分岐で、オブジェクトの同一性を確認する際に使用します。

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

if a is b:
    print("a and b are the same object")
    
if a is not c:
    print("a and c are not the same object")
```

以下は、正規表現を使用した条件分岐で `is` を活用する例です。  
正規表現のマッチ結果が `None`（マッチしない）かどうかを判定する際に、`is` 演算子を使用することができます。

```python
import re

# 検索対象の文字列
text = "Hello, World!"

# 正規表現で "Hello" を検索
match = re.search(r"Hello", text)

if match is not None:
    print("Pattern found:", match.group())
else:
    print("Pattern not found")
```

```python
# 正規表現で "Python" を検索
match = re.search(r"Python", text)

if match is None:
    print("Pattern not found")
else:
    print("Pattern found:", match.group())
```


---
# in
`in` 演算子を使うと、要素がシーケンス（リスト、タプル、文字列など）に含まれているかを判定できます。  
条件分岐で、要素の存在を確認する際に使用します。

```python
fruits = ["apple", "banana", "cherry"]

if "banana" in fruits:
    print("banana is in the list")
    
if "orange" not in fruits:
    print("orange is not in the list")
```
