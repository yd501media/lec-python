# **Package Management with pip**
`pip`は、Python公式のパッケージ管理ツールで、外部ライブラリのインストール、アップデート、アンインストールが簡単に行えます。  
`requirements.txt`を使用して依存関係を一括管理することも可能です。


---
# pipの概要
`pip` はPythonの公式パッケージ管理ツールであり、PyPI（Python Package Index）から  
パッケージをインストール、アンインストール、管理するために使用されます。  
Pythonプロジェクトで外部ライブラリを簡単に利用できるようにする必須のツールです。


---
# パッケージのインストール
`pip install` コマンドを使用して、必要なパッケージをPyPIからインストールします。  
依存関係がある場合、自動的に関連するパッケージもインストールされます。

以下は `numpy` をインストールする例です。
```bash
pip install numpy
```

結果は以下のように出力されます。
```bash
Collecting numpy
  Downloading numpy-1.23.5-cp39-cp39-win_amd64.whl (12.3 MB)
Installing collected packages: numpy
Successfully installed numpy-1.23.5
```


---
# インストール済みパッケージの確認
`pip list` コマンドを使用して、現在インストール済みのパッケージ一覧を表示します。  
バージョン情報も含まれるため、パッケージのアップデートが必要かを確認できます。

以下は、`pip list` コマンドを実行した結果の例です。
```bash
Package         Version
--------------- --------
numpy           1.23.5
pandas          1.5.3
requests        2.28.1
pip             24.2
```


---
# パッケージのアンインストール
pip uninstall コマンドを使用して、不要になったパッケージをシステムから削除します。  
複数のパッケージを同時にアンインストールすることも可能です。

以下は `numpy` をアンインストールする例です。
```bash
pip uninstall numpy
```

上記コマンドを実行すると以下のように出力され、  
最終行に `Proceed (Y/n)?` と表示され、アンインストールを進めるか確認されます。
```bash
Found existing installation: numpy 1.23.5
Uninstalling numpy-1.23.5:
  Would remove:
    /path/to/python/site-packages/numpy/*
    /path/to/python/site-packages/numpy-1.23.5.dist-info/*
Proceed (Y/n)?
```

`Y` もしくは `y` を入力するとアンインストールが開始されます。  
以下が出力されれば、アンインストールは完了です。
```bash
Successfully uninstalled numpy-1.23.5
```

`pip list` を実行して、出力に `numpy` が含まれていないことを確認します。
```bash
Package         Version
--------------- --------
pandas          1.5.3
requests        2.28.1
pip             24.2
```


---
# 仮想環境での利用
`pip` は仮想環境内で使用されることで、プロジェクトごとに独立したパッケージ管理を可能にします。  
これにより、複数プロジェクト間でのパッケージ競合を防ぎます。

仮想環境を起動します。仮想環境が作成されていない場合は作成します。

パッケージのインストール、インストール済みぱケージの確認、パッケージのアンインストールの方法は上記と同様です。


---
# requirements.txtの作成
`pip freeze` コマンドは、現在のPython環境または仮想環境にインストールされているパッケージとそのバージョンを固定形式で表示します。  
この出力は、`requirements.txt` ファイルに保存して他の環境で再現するために使用されます。


以下のコマンドを実行することで `requirements.txt` を作成します。
```bash
pip freeze > requirements.txt
```

出力された `requirements.txt` は以下の形式です。
```planetext
numpy==1.23.5
pandas==1.5.3
requests==2.28.1
```


---
# requirements.txtの利用
`requirements.txt` ファイルは、プロジェクトで必要なパッケージのリストを記載したファイルです。  
`pip install -r requirements.txt` を実行することで、ファイルに記載されたパッケージを一括インストールできます。

結果は以下のように出力されます。
```bash
Collecting numpy==1.23.5
  Downloading numpy-1.23.5-cp39-cp39-win_amd64.whl (12.3 MB)
Collecting pandas==1.5.3
  Downloading pandas-1.5.3-cp39-cp39-win_amd64.whl (11.2 MB)
Collecting requests==2.28.1
  Downloading requests-2.28.1-py3-none-any.whl (62 kB)
Installing collected packages: numpy, pandas, requests
Successfully installed numpy-1.23.5 pandas-1.5.3 requests-2.28.1
```


---
# pipのアップグレード
`pip install`, `pip list` コマンドなどを実行した際に、`pip` のバージョンが古い場合、  
以下のような `Warning` が表示されることがあります。
```bash
WARNING: You are using pip version 23.2.1; however, version 24.2 is available.
You should consider upgrading via the 'python -m pip install --upgrade pip' command.
```

この場合は以下のコマンドを実行して、`pip` 自体を最新バージョンに更新します。
```bash
python -m pip install --upgrade pip
```

結果は以下のように出力されます。
```bash
Collecting pip
  Downloading pip-24.2-py3-none-any.whl (2.2 MB)
     |████████████████████████████████| 2.2 MB 5.2 MB/s
Installing collected packages: pip
  Attempting uninstall: pip
    Found existing installation: pip 23.2.1
    Uninstalling pip-23.2.1:
      Successfully uninstalled pip-23.2.1
Successfully installed pip-24.2
```
