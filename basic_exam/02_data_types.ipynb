# 数値
Pythonには数を扱う概念として、整数を表す`int`型と実数を表す`float`型が存在しています。
## int
小数点を含まない数値を扱うデータ型。整数型。

````python
print(0)
print(type(0))

print(100)
print(type(100))

print(-255)
print(type(-255))
````

PythonではPCのメモリに収まる範囲で、値範囲の制限がありません。  
例えばC言語のint型は `32ビット` で表現され、範囲は `-2**31` ~ `2**31-1` に制限されます。  
- 最小値: `-2,147,483,648`
- 最大値: `+2,147,483,647`

参考
- [IBM® Informix® 12.10 > 整数 (INTEGER) 型](https://www.ibm.com/docs/ja/informix-servers/12.10?topic=types-integer-data-type)
- [Microsoft Learn Challenge > データ型の範囲](https://learn.microsoft.com/ja-jp/cpp/cpp/data-type-ranges?view=msvc-170)

Pythonではこの値範囲を超えて値を扱うことができます。

````python
# 例えば、上記の最小値、最大値をそれぞれ2乗する
print((-2 ** 31) ** 2)
print((2 ** 31 - 1) ** 2)
````

## float
小数点を含む数値を扱うデータ型。浮動小数点型。

````python
print(0.0)
print(type(0.0))

print(3.14)
print(type(3.14))

print(-0.5)
print(type(-0.5))
````

浮動小数点型は内部的にバイナリ表現を使用するため、計算結果に**誤差が発生**することがあります。  

````python
print(0.1 + 0.2)
````

浮動小数点型では、非常に大きな値や非常に小さな値を扱う際に**科学記数法**を使用できます。

````python
print(1.23e4)    # 1.23 × 10^4
print(type(1.23e4))

print(4.56e-3)   # 4.56 × 10^-3
print(type(4.56e-3))
````

# 文字列
Pythonには文字列を扱うデータ型として、`str`型が存在します。  
文字列はシングルクォート（`''`）またはダブルクォート（`""`）で定義します。

## 'single quote' と "double quote"
文字列を扱う場合、どちらを使用しても問題ありません。

````python
print('Hello, world!')
print("Hello, world!")
````

例えば、文字列の中に'(single quote)や"(double quote)を使用したい場合は、以下のように'と"を使い分ける。

````python
# 文字列の中に"(double quote)が出現する場合は
# 文字列の定義に'(single quote)を使用する
print('He said, "Hello, World!"')

# 文字列の中に'(single quote)が出現する場合は
# 文字列の定義に"(double quote)を使用する
print("It's a beautiful day.")
````

## raw文字列
raw文字列とは、バックスラッシュをエスケープせず、そのまま扱う文字列です。  
文字列を定義する際に、`""` または `''` の前に `r` を付与することで使用できます。  
  
`\`(バックスラッシュ)は通常エスケープシーケンスとして扱われますが、raw文字列を使うことでそのままの形式を保持できます。  
Windows環境などでは`\`が`￥`(エンマーク)として表示されることがあります。  

正規表現やWindowsのファイルパスなど、`\` を多用するケースで特に便利です。  
例えば、Windows環境でファイルパスを扱う場合、raw文字列を使用しないとエラーとなる場合があります。

````python
print("C:\Users\UserName\Documents\Python\dev.py")
````

raw文字列を使用しない場合は以下のようになります。

````python
print("C:\\Users\\UserName\\Documents\\Python\\dev.py")
````

しかし、raw文字列を使用すれば以下のようにシンプルに書き表すことができます。

````python
print(r"C:\Users\UserName\Documents\Python\dev.py")
````

以下のように正規表現のパターンにraw文字列を使用することができます。

````python
import re
print(re.search(r"\D{6}\d+", "hLtIzANZBB8ENVmzSvk6vukChP4szrxoba1KmS9YoHy7bHjjNJ"))
````

## f文字列 (フォーマット済み文字列)
f文字列は、文字列内に変数や式を埋め込んで簡単に文字列を生成できる便利な方法です。  
文字列を定義する際に、`""` または `''` の前に `f` を付与することで使用できます。  

````python
print(f"1 + 2 = {1 + 2}")
print(f"Pi is nearly equal to {3.14159:0.2f} .")
````

# 変数の宣言
Pythonでは、変数名は 小文字で定義 するのが一般的です（PEP 8のスタイルガイドに従う）。  
変数は、値を格納するために使用します。

````python
# 変数の宣言
name = "Alice"       # 名前を格納
age = 30             # 年齢を格納
is_employed = True   # 就業中であるかの真偽値を格納
job_title = "Engineer"  # 職業名を格納

# 変数の使用
print("Name:", name)
print("Age:", age)
print("Is Employed:", is_employed)
print("Job Title:", job_title)
````

#### **注意事項**
1. **変数名のルール**:
   - 変数名は文字、数字、アンダースコア（`_`）を使用可能。
   - 数字で始めることはできない（例: `1var` は無効）。
   - Pythonの予約語（例: `if`, `for`, `while` など）は使用できない。

2. **変数名のスタイル**:
   - **小文字を使用**: `name`, `age`（一般的な変数）。
   - **アンダースコアで区切る**: `job_title`, `is_employed`（読みやすさの向上）。


# 定数の宣言
Pythonでは、定数を厳密に定義する構文はありませんが、 **全て大文字の変数名を使う** ことで、  
「これは定数として扱うべき値」であることを明示するのが一般的です（PEP 8のスタイルガイドに従う）。  
定数は通常、変更しない値を表すために使用します。

````python
# 定数の宣言
PI = 3.14159         # 円周率
MAX_CONNECTIONS = 100  # 最大接続数
DEFAULT_TIMEOUT = 60   # デフォルトのタイムアウト（秒）

# 定数の使用
print("The value of PI is:", PI)
print("Maximum connections allowed:", MAX_CONNECTIONS)
print("Default timeout is:", DEFAULT_TIMEOUT)
````

#### **注意事項**
1. **定数名のルール**:
   - 定数名はすべて大文字で記述し、単語はアンダースコア（`_`）で区切る。
      ```python
      GOOD_CONSTANT = 42  # 良い例
      badConstant = 42    # 悪い例
      ```
2. **定数は変更しない**:
   - Pythonでは定数を変更可能ですが、意図的に変更しないようにすることが重要。
      ```python
      PI = 3.14159
      PI = 3.0  # 推奨されない（意図が崩れる）
      ```

# 型変換
Pythonでは、異なるデータ型間での変換が可能です。  
`int()`, `float()`, `str()` などの組み込み関数を使用して、数値や文字列を相互に変換できます。  
  
以下は`int`から`float`への変換の例です。

````python
x_int = 42
print(f"{x_int=}, {type(x_int)=}")

x_float = float(x_int)
print(f"{x_float=}, {type(x_float)=}")
````

以下は `float` から `int` への変換の例です。  
`float` を `int` に変換すると、小数点以下の部分は切り捨てられます。

````python
p_float = 3.14
print(f"{p_float=}, {type(p_float)=}")

p_int = int(p_float)
print(f"{p_int=}, {type(p_int)=}")
````

以下は`str`から`float`への変換の例です。

````python
s_str = "3.14159"
print(f"{s_str=}, {type(s_str)=}")

s_float = float(s_str)
print(f"{s_float=}, {type(s_float)=}")
````

文字列として浮動小数点数を用意した場合、これを`int`に変換することはできません。

````python
s_int = int(s_str)
````

文字列として整数を用意した場合、これは`int`にも`float`に変換することができます。

````python
s_str = "42"
print(f"{s_str=}, {type(s_str)=}")

s_int = int(s_str)
print(f"{s_int=}, {type(s_int)=}")

s_float = float(s_str)
print(f"{s_float=}, {type(s_float)=}")
````

`int`もしくは`float`は`str`に変換可能です。

````python
print(f"{x_int=}, {type(x_int)=}")
x_str = str(x_int)
print(f"{x_str=}, {type(x_str)=}", end="\n\n")

print(f"{p_float=}, {type(p_float)=}")
p_str = str(p_float)
print(f"{p_str=}, {type(p_str)=}")
````

# 数値演算
Pythonでは、基本的な数値演算をサポートしており、簡単な計算から複雑な計算まで幅広く利用できます。

### 加算 (`+`)
2つの数を足します。

````python
# int同士の加算結果はint
result = 5 + 3
print(f"{result=}, {type(result)=}")

# intとfloatの加算結果はfloat
result = 5 + 3.0
print(f"{result=}, {type(result)=}")

# floatとfloatの加算結果はfloat
result = 5.1 + 3.2
print(f"{result=}, {type(result)=}")
````

### 減算 (`-`)
2つの数を引きます。

````python
# int同士の減算結果はint
result = 5 - 3
print(f"{result=}, {type(result)=}")

# intとfloatの減算結果はfloat
result = 5 - 3.0
print(f"{result=}, {type(result)=}")

# floatとfloatの減算結果はfloat
result = 5.1 - 3.2
print(f"{result=}, {type(result)=}")
````

### 乗算 (`*`) 
2つの数を掛けます。

````python
# int同士の乗算結果はint
result = 5 * 3
print(f"{result=}, {type(result)=}")

# intとfloatの乗算結果はfloat
result = 5 * 3.0
print(f"{result=}, {type(result)=}")

# floatとfloatの乗算結果はfloat
result = 5.1 * 3.2
print(f"{result=}, {type(result)=}")
````

### 除算 (`/`)
2つの数を割ります（結果は常に浮動小数点数）。

````python
result = 5 / 2
print(f"{result=}, {type(result)=}")
````

### 整数除算 (`//`)
2つの数を割り、小数点以下を切り捨てます。

````python
result = 5 // 2
print(f"{result=}, {type(result)=}")
````

### 剰余 (`%`)
割り算の余りを求めます。

````python
result = 5 % 2
print(f"{result=}, {type(result)=}")
````

### べき乗 (`**`)
指定した数のべき乗を計算します。

````python
result = 5 ** 3
print(f"{result=}, {type(result)=}")
````

# 文字列操作
Pythonでは、文字列に対してさまざまな操作が可能です。  
文字列を効率的に操作するための基本的な機能を紹介します。

## 結合
複数の文字列を連結して1つの文字列にまとめます。

````python
result = "Hello" + ", " + "World" + "!"
print(result)
````

## スライス
文字列の一部を取得します。インデックスを指定して部分文字列を切り出します。

````python
text = "Python"
sliced = text[0:3]
print(sliced)
````

## メソッド
文字列を操作するための便利なメソッドが豊富に用意されています。以下は例です。  
- **`upper()`**: 文字列をすべて大文字に変換。
- **`find()`**: 部分文字列の最初の出現位置を返す。見つからない場合は `-1` を返す。
- **`replace()`**: 部分文字列を別の文字列に置き換える。
- **`split()`**: 指定された区切り文字で分割し、リストを返す。
- **`join()`**: 指定した文字列でリストなどを連結。
- **`islower()`**: 小文字のみで構成されているか確認。

### upper()

````python
s = "hello"
print(f"{s=} -> {s.upper()}")
````

### find()

````python
s = "hLtIzANZBB8ENVmzSvk6vukChP4szrxoba1KmS9YoHy7bHjjNJ"
print(f"{s=} -> {s.find("szrxoba")}")
print(f"{s=} -> {s.find("abcdefg")}")
````

### replace()

````python
s = "Hello, World!"
print(f"{s=} -> {s.replace("World", "Python")}")
````

### split()

````python
s = "apple,banana,orange"
print(f"{s=} -> {s.split(",")}")
````

### join()

````python
l = ['apple', 'banana', 'orange']
print(f"{l=} -> {'_'.join(l)}")
````

### islower()

````python
s = "Hello, world!"
print(f"{s=} -> {s.islower()=}")

s = "hello, world!"
print(f"{s=} -> {s.islower()=}")
````
